## 2. The question: decision, action, cost of a wrong call

## Research Question

**Which content pages should be prioritized for human review based on signals such as visibility, decline, freshness, and possible opportunity?**

## Unit of Analysis

The unit of analysis is **one content page**.

Each row in the dataset represents one content page with different measurements related to its performance.

## Decision

The main decision is:

**Which pages should a human reviewer check first when there is limited time or review capacity?**

## Action

The results can help a reviewer decide what to do with a page after checking it.

Possible actions can include:

- Refresh or update the content
- Expand or improve the content
- Protect and monitor an important page
- Check for possible CTR or engagement problems
- Continue monitoring the page without taking action

The system will not automatically make these changes. Its purpose is to help rank the pages and decide which ones should be reviewed first.

## Cost of a Wrong Recommendation

A false positive means that a page is recommended for review even though it may not really need attention. This can waste the team's time and resources.

A false negative means that an important page is not recommended, even though it may have a problem or a good opportunity. This could cause the team to miss a page that should have been reviewed.

Because the team may have limited time, both types of mistakes are important. For this reason, the system should be used to support human decisions, not to make decisions automatically.

## 4. Careful words: what I can and can't claim

At this stage, I cannot say that a machine learning model will definitely identify the best pages to refresh.

I also cannot say that refreshing a page will directly cause traffic, visibility, or engagement to improve. The available data is observational, so it can show patterns and relationships, but it cannot prove causation by itself.

If this project creates a useful ranking, I can say that the system may help prioritize pages for human review based on the available data and the results of the evaluation.

A page with a high score does not mean that the page will definitely improve after being updated. It only means that the page may be worth reviewing.

During the project, I will also need to check for possible problems such as noisy data, low-volume pages, seasonality, and data leakage before making stronger conclusions.

I will use careful language such as:

- "The data suggests..."
- "The model identified..."
- "This ranking may help prioritize..."
- "There is an observed relationship between..."

I will avoid statements such as:

- "This proves..."
- "This caused..."
- "This guarantees improvement..."
- "The model knows which pages will succeed..."
