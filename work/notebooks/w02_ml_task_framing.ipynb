## 5. Why ML beats a fixed rule here

A fixed rule (e.g., IF age > 365 AND impressions < 100 THEN refresh) is too rigid and messy for this problem.
It ignores nuance: A page might be only 6 months old but experienced a sudden 80% traffic drop due to an algorithm update. A fixed rule would ignore it, but it's a high-priority rescue candidate.
It cannot weigh multiple factors: ML can learn non-linear interactions between features (e.g., a moderate traffic drop on a page with historically very high impressions is more critical than a severe drop on a page that never had traffic).
Therefore, ML acts as a dynamic decision-support system, not a static filter.
